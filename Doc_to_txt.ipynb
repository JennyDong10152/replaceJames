```python
import os
from docx import Document

# Function to extract text, chart descriptions, and tables from a DOCX file
def extract_text_with_charts(docx_path, output_txt_path, image_folder):
    doc = Document(docx_path)
    
    if not os.path.exists(image_folder):
        os.makedirs(image_folder)
    
    with open(output_txt_path, "w", encoding="utf-8") as txt_file:
        txt_file.write("Extracted Text from DOCX\n\n")
        
        for para in doc.paragraphs:
            if para.text.strip():  # Ignore empty paragraphs
                if "chart" in para.text.lower() or "graph" in para.text.lower():
                    txt_file.write(f"[CHART DESCRIPTION] {para.text}\n")
                else:
                    txt_file.write(f"{para.text}\n")
        
        txt_file.write("\nExtracted Tables:\n\n")
        for table in doc.tables:
            txt_file.write("[TABLE DATA]\n")
            for row in table.rows:
                row_text = " | ".join(cell.text.strip() for cell in row.cells)
                txt_file.write(row_text + "\n")
            txt_file.write("\n")
    
    print(f"Text saved to {output_txt_path}")

# Function to extract images (charts/graphs) from the DOCX file
def extract_images(docx_path, image_folder):
    doc = Document(docx_path)
    count = 0
    
    for rel in doc.part.rels:
        if "image" in doc.part.rels[rel].target_ref:
            count += 1
            image = doc.part.rels[rel].target_part.blob
            image_path = os.path.join(image_folder, f"chart_{count}.png")
            with open(image_path, "wb") as img_file:
                img_file.write(image)
    
    print(f"Extracted {count} images to {image_folder}")

# Example Usage
docx_file = "input.docx"  # Replace with your DOCX file path
output_text_file = "output.txt"
image_output_folder = "charts_folder"

extract_text_with_charts(docx_file, output_text_file, image_output_folder)
extract_images(docx_file, image_output_folder)
```
